# XGBoost Regressor

```python
from xgboost import XGBRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

```python
xgb_reg = XGBRegressor(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=5,
    random_state=42
)

xgb_reg.fit(X_train1, Y_train1)
```

```python
Y_pred_xgb_reg = xgb_reg.predict(X_test1)
```

```python
print("MAE:", mean_absolute_error(Y_test1, Y_pred_xgb_reg))
print("MSE:", mean_squared_error(Y_test1, Y_pred_xgb_reg))
print("RMSE:", np.sqrt(mean_squared_error(Y_test1, Y_pred_xgb_reg)))
print("R2 Score:", r2_score(Y_test1, Y_pred_xgb_reg))
```

# Gradient Boosting Regressor
